# 03 Semi-synthetic Benchmark — Controlled Observation-Policy Shift
Generates latent trajectories + policy-dependent observation; tests oracle invariance and held-out policy shift (spec §8).

```
import numpy as np
from vist_nrf.benchmarks.semisynthetic import SemiSyntheticConfig, generate_all, oracle_predict
from vist_nrf.eval.metrics import auroc

cfg = SemiSyntheticConfig(n_patients=100, n_hours=48, n_vars=8, n_envs=2, seed=0,
                          w=[0.5, 0.8], r=[0.2, 0.2])
data = generate_all(cfg, seed=0)
scores, truths = oracle_predict(data)
y = np.array([truths[k] for k in scores])
s = np.array([scores[k] for k in scores])
print('Oracle AUROC (policy-invariant upper bound): %.3f' % auroc(y, s))
print('Patients per env:', {e: len(p) for e, p in data.items()})
```
